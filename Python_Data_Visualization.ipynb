```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ------------------------------------------
# 1. Create Sample Dataset
# ------------------------------------------

data = {
    "Student": ["Deep", "Monil", "Kriya", "Siya", "Karan", "Amit", "Sneha"],
    "Maths": [95, 76, 92, 78, 99, 91, 87],
    "Science": [86, 80, 85, 90, 81, 89, 86],
    "English": [78, 74, 60, 87, 75, 78, 80]
}

df = pd.DataFrame(data)
print("Student Dataset")
print(df)
```

```python
print("\nShape :", df.shape)
```

```python
print("\nColumns")
print(df.columns)
```

```python
print("\nSummary Statistics")
print(df.describe())
```

```python
plt.figure(figsize=(8,5))
plt.plot(df["Student"], df["Maths"], marker='o')
plt.title("Maths Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.bar(df["Student"], df["Science"])
plt.title("Science Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Maths"], df["Science"])
plt.title("Maths vs Science")
plt.xlabel("Maths")
plt.ylabel("Science")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(7,5))
plt.hist(df["English"], bins=5)
plt.title("English Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(6,6))
plt.pie(
    df["Maths"],
    labels=df["Student"],
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Maths Marks Percentage")
plt.show()
```

```python
plt.figure(figsize=(6,5))
plt.boxplot(df["Science"])
plt.title("Science Marks Box Plot")
plt.ylabel("Marks")
plt.show()
```

```python
sns.lineplot(x="Student", y="English", data=df)

plt.title("English Marks")
plt.show()
```

```python
sns.barplot(x="Student", y="Maths", data=df)

plt.title("Maths Bar Plot")
plt.show()
```

```python
sns.scatterplot(
    x="Maths",
    y="Science",
    data=df
)

plt.title("Maths vs Science")
plt.show()
```

```python
sns.histplot(df["Maths"], bins=5)

plt.title("Maths Histogram")
plt.show()
```

```python
sns.histplot(df["Maths"], bins=5)

plt.title("Maths Histogram")
plt.show()
```

```python
sns.heatmap(
    df[["Maths","Science","English"]].corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
sns.pairplot(
    df[["Maths","Science","English"]]
)

plt.show()
```

```python
df["Average"] = (
    df["Maths"] +
    df["Science"] +
    df["English"]
) / 3

print("\nAverage Marks")
print(df[["Student","Average"]])
```
