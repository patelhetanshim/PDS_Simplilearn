```python
import pandas as pd

# Create a list of numbers from 0 to 4
l = [x for x in range(5)]
s = pd.Series(l)
print(s)
```

```python
# Access element by position index
print(s[3])
```

```python
# Recreate the Series with custom string labels as index
s = pd.Series(l, index=['a', 'b', 'c', 'd', 'e'])
print(s)
```

```python
# Access element using its custom string label
print(s['e'])
```

```python
# Recreate the Series with duplicate index labels ('d')
s = pd.Series(l, index=['a', 'b', 'c', 'd', 'd'])
```

```python
# Accessing a duplicate label returns all matching elements
print(s['d'])
```
